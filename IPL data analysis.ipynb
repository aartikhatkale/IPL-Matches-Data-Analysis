```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
matches = pd.read_csv(r"C:\Users\user\Downloads\matches.csv")
deliveries = pd.read_csv(r"C:\Users\user\Downloads\deliveries.csv")
```

```python
#Display first 5 rows
display(matches.head())
display(deliveries.head())
```

```python
#Dataset information
matches.info()
deliveries.info()
```

```python
#Dataset shape
print("Matches Shape:", matches.shape)
print("Deliveries Shape:", deliveries.shape)
```

```python
#Missing values
print(matches.isnull().sum())
print(deliveries.isnull().sum())
```

```python
#Statistical summary
print(matches.describe())
print(deliveries.describe())
```

```python
#Total matches
print("Total Matches:",matches.shape[0])
```

```python
#Teams most wins
print("Team with Most Wins:",matches['winner'].value_counts().idxmax())
print("Number of Wins:",matches['winner'].value_counts().max())
```

```python
#Top 10 Winning Teams
top_teams=matches['winner'].value_counts().head(10)
plt.figure(figsize=(10,5))
top_teams.plot(kind='bar')
plt.title("Top 10 Winning Teams")
plt.xlabel("Teams")
plt.ylabel("Wins")
plt.xticks(rotation=45)
plt.show()           
```

```python
#Matches Per Seasons
season_matches=matches['season'].value_counts().sort_index()
plt.figure(figsize=(10,5))
season_matches.plot(kind='bar')
plt.title("Matches Per Season")
plt.xlabel("Season")
plt.ylabel("Matches")
plt.show()
```

```python
#Toss Winners
toss=matches['toss_winner'].value_counts().head(10)
plt.figure(figsize=(10,5))
toss.plot(kind='bar')
plt.title("Top Toss Winners")
plt.xlabel("Teams")
plt.ylabel("Toss Wins")
plt.xticks(rotation=45)
plt.show()
```

```python
#Toss Decision
matches['toss_decision'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title("Toss Decision")
plt.ylabel("")
plt.show()
```

```python
#Top player of the match
pom=matches['player_of_match'].value_counts().head(10)
plt.figure(figsize=(10,5))
pom.plot(kind='bar')
plt.title("Top Player of the Match")
plt.xticks(rotation=45)
plt.show()
```

```python
#Top Run scoreres(Orange cap)
top_runs = deliveries.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_runs.plot(kind='bar')
plt.title("Top 10 Run Scorers")
plt.xlabel("Batsman")
plt.ylabel("Runs")
plt.xticks(rotation=45)
plt.show()
```

```python
#Top Wicket takers(Purple cap)
# Filter wickets
wickets = deliveries[deliveries['player_dismissed'].notna()]

# Count wickets by bowler
top_wickets = wickets['bowler'].value_counts().head(10)

print(top_wickets)

plt.figure(figsize=(10,5))
top_wickets.plot(kind='bar', color='purple')
plt.title("Top 10 Wicket Takers (Purple Cap)")
plt.xlabel("Bowler")
plt.ylabel("Wickets")
plt.xticks(rotation=45)
plt.show()
```

```python
#Top venues
venue = matches['venue'].value_counts().head(10)
plt.figure(figsize=(10,5))
venue.plot(kind='bar')
plt.title("Top Venues")
plt.xticks(rotation=90)
plt.show()
```

```python
#Matches won by runs
plt.figure(figsize=(10,5))
matches['win_by_runs'].hist(bins=20)
plt.title("Distribution of Win by Runs")
plt.show()
```

```python
#Matches won by wickets
plt.figure(figsize=(10,5))
matches['win_by_wickets'].hist(bins=10)
plt.title("Distribution of Win by Wickets")
plt.show()
```
